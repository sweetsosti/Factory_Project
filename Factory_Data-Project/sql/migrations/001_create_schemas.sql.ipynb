```python
%sql

DESCRIBE TABLE factory_catalog.bronze.machines;
```

```python
%sql

SELECT *
FROM factory_catalog.bronze.machines
LIMIT 10;
```

```python
%sql

SELECT
    machine_id,
    COUNT(*) AS record_count
FROM factory_catalog.bronze.machines
GROUP BY machine_id
HAVING COUNT(*) > 1;
```

```python
%sql

SELECT
    COUNT(*) AS total_records,
    SUM(CASE WHEN machine_id IS NULL OR TRIM(machine_id) = '' THEN 1 ELSE 0 END) AS missing_machine_id,
    SUM(CASE WHEN machine_name IS NULL OR TRIM(machine_name) = '' THEN 1 ELSE 0 END) AS missing_machine_name,
    SUM(CASE WHEN production_line IS NULL OR TRIM(production_line) = '' THEN 1 ELSE 0 END) AS missing_production_line,
    SUM(CASE WHEN machine_type IS NULL OR TRIM(machine_type) = '' THEN 1 ELSE 0 END) AS missing_machine_type,
    SUM(CASE WHEN install_date IS NULL OR TRIM(install_date) = '' THEN 1 ELSE 0 END) AS missing_install_date,
    SUM(CASE WHEN rated_cycle_time_seconds IS NULL THEN 1 ELSE 0 END) AS missing_cycle_time
FROM factory_catalog.bronze.machines;
```

```python
%sql

SELECT
    machine_id,
    COUNT(*) AS record_count
FROM factory_catalog.bronze.machines
GROUP BY machine_id
HAVING COUNT(*) > 1;
```

```python
%sql

SELECT *
FROM factory_catalog.bronze.machines
WHERE rated_cycle_time_seconds IS NULL
   OR rated_cycle_time_seconds <= 0;
```
