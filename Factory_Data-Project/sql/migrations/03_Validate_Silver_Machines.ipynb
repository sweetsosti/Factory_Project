```python
%sql

SELECT
    'Bronze' AS layer,
    COUNT(*) AS total_records
FROM factory_catalog.bronze.machines

UNION ALL

SELECT
    'Silver' AS layer,
    COUNT(*) AS total_records
FROM factory_catalog.silver.machines;
```

```python
%sql

SELECT *
FROM factory_catalog.silver.machines
WHERE install_date IS NULL
   OR rated_cycle_time_seconds IS NULL
   OR rated_cycle_time_seconds <= 0;
```
