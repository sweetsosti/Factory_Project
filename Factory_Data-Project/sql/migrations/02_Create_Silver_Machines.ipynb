```python
%sql

CREATE TABLE IF NOT EXISTS factory_catalog.silver.machines
USING DELTA
AS
SELECT DISTINCT
    TRIM(machine_id) AS machine_id,
    TRIM(machine_name) AS machine_name,
    TRIM(production_line) AS production_line,
    TRIM(machine_type) AS machine_type,
    TRY_CAST(install_date AS DATE) AS install_date,
    rated_cycle_time_seconds
FROM factory_catalog.bronze.machines
WHERE machine_id IS NOT NULL
  AND TRIM(machine_id) <> ''
  AND machine_name IS NOT NULL
  AND TRIM(machine_name) <> ''
  AND production_line IS NOT NULL
  AND TRIM(production_line) <> ''
  AND machine_type IS NOT NULL
  AND TRIM(machine_type) <> ''
  AND TRY_CAST(install_date AS DATE) IS NOT NULL
  AND rated_cycle_time_seconds > 0;
```

```python
%sql

SELECT *
FROM factory_catalog.silver.machines
LIMIT 10;
```

```python
%sql

SELECT COUNT(*) AS silver_records
FROM factory_catalog.silver.machines;
```

```python
%sql

SELECT
    machine_id,
    COUNT(*) AS record_count
FROM factory_catalog.silver.machines
GROUP BY machine_id
HAVING COUNT(*) > 1;
```
