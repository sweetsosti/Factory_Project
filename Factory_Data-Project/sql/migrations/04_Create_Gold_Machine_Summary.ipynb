```python
%sql

CREATE TABLE IF NOT EXISTS factory_catalog.gold.machine_summary
USING DELTA
AS
SELECT
    production_line,
    COUNT(*) AS total_machines,
    COUNT(DISTINCT machine_type) AS machine_types,
    ROUND(AVG(rated_cycle_time_seconds), 2)
        AS avg_cycle_time_seconds,
    MIN(rated_cycle_time_seconds) AS min_cycle_time_seconds,
    MAX(rated_cycle_time_seconds) AS max_cycle_time_seconds
FROM factory_catalog.silver.machines
GROUP BY production_line;
```

```python
%sql

SELECT *
FROM factory_catalog.gold.machine_summary
ORDER BY production_line;
```
